```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('course_lead_scoring_2026.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.dtypes
```

```python
categorical_columns = list(df.dtypes[df.dtypes == 'str'].index)
numerical_columns = list(df.dtypes[df.dtypes != 'str'].index)

numerical_columns
```

```python
df_cat = df[categorical_columns].fillna('NA')
df_cat.isnull().sum()
```

```python
df_num = df[numerical_columns].fillna(0.0)
df_num.isnull().sum()
```

```python
for cat in categorical_columns:
    df[cat] = df_cat[cat]

for num in numerical_columns:
    df[num] = df_num[num]

df.head()
```
