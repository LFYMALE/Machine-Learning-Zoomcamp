```python
import numpy as np
import pandas as pd

df = pd.read_csv('car_fuel_efficiency_2026.csv')

df.describe()
# 2-The Records count is 10000

n = df['fuel_type'].nunique()

print(n)
# 3-There are 3 fuel type

print(df.isnull().sum())
# 4-The number of missing values is 2 (horsepower, acceleration)

m = df['fuel_efficiency_mpg'].max()
print(m)
# 5-Max fuel efficiency is 41.2

# 6-1-Find the median value of the horsepower column in the dataset.
med = df['horsepower'].median()
print(f"Median: {med}")

# 6-2-Next, calculate the most frequent value of the same horsepower column.
most_value = df['horsepower'].mode()[0]
print(f'The most frequent value is: {most_value}')

# 6-3-Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
df['horsepower'] = df['horsepower'].fillna(most_value)

# 6-3-Now, calculate the median value of horsepower once again.
new_median = df['horsepower'].median()
print(f'Previous median: {med}')
print(f'New median: {new_median}')


# 7-1-Select all the cars from Asia
cars_asia = df[df['origin'] == 'Asia']
# 7-2-Select only columns vehicle_weight and model_year
vehWeight_modelYear = cars_asia[['vehicle_weight', 'model_year']]
#print(vehWeight_modelYear)

# 7-3-Select the first 7 values
first7 = vehWeight_modelYear[:7]
# print(first7)

# 7-4-Get the underlying NumPy array. Let's call it X.
X = first7.to_numpy()
# print(X)

# 7-5-Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX
XTX = X.T.dot(X)
# print(XTX)

# 7-6-Invert XTX
inv_XTX = np.linalg.inv(XTX)
# print(inv_XTX)

# 7-7-Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200]
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
# print(y)

# 7-8-Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w

w_ = inv_XTX.dot(X.T)
w = w_.dot(y)
print(w.sum())
```
